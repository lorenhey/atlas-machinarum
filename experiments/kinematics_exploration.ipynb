# Kinematics Exploration
This notebook demonstrates how to interact with the Atlas Machinarum API programmatically to compare mechanisms.

```python
from atlas_machinarum.library import SliderCrank, Chebyshev
import matplotlib.pyplot as plt
import numpy as np

# 1. Slider Crank Analysis
sc = SliderCrank(crank=0.05, rod=0.20, offset=0.0)
theta_range = np.linspace(0, 2*np.pi, 100)
path = sc.path(theta_range)

slider_x = [state.points['B'][0] for state in path]
plt.plot(theta_range, slider_x, label='Slider X')
plt.title('Slider Crank: Position vs Crank Angle')
plt.xlabel('Angle (rad)')
plt.ylabel('X Position (m)')
plt.legend()
plt.show()
```

```python
# 2. Chebyshev Straight Line Analysis
cheb = Chebyshev(a=1.0)
cheb_path = cheb.path(theta_range)

p_x = [state.points['P'][0] for state in cheb_path]
p_y = [state.points['P'][1] for state in cheb_path]

plt.plot(p_x, p_y, label='Coupler Point P')
plt.title('Chebyshev Linkage: Coupler Curve')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal')
plt.legend()
plt.show()
```
